### Server - GUI Client Structure
- Server: flask app to listen
- Client: GUI class to make structure & communicate with server

```python
""" SERVER """
import os
from flask import Flask, jsonify, request, send_file
import numpy as np

# correct_image = "./correct_image.jpg"
usertracking_image = "./usertracking_image.jpg"
patrol_image = "./patrol_image.jpg"
map_image = "./map_image.jpg"
error_image = "./error_image.jpg"

app = Flask(__name__)

@app.route(f'/GUI/stt')
def server_stt_output():
    # Ensure the file exists before trying to send it
    output_text = "This is a sample STT output from the server."
    return jsonify({"output": output_text})

@app.route(f'/GUI/tts')
def server_tts_output():
    # Ensure the file exists before trying to send it
    output_text = "This is a sample TTS output from the server."
    return jsonify({"output": output_text})

# ========================================================================
@app.route('/GUI/vlm')
def server_vlm_output():
    if os.path.exists(patrol_image):
        return send_file(patrol_image, mimetype="image/jpg")
    else:
        return send_file(error_image, mimetype="image/jpg"), 404
# ========================================================================

@app.route('/GUI/usertracking')
def server_usertracking_output():
    if os.path.exists(usertracking_image):
        return send_file(usertracking_image, mimetype="image/jpg")
    else:
        return send_file(error_image, mimetype="image/jpg"), 404
    
@app.route('/GUI/patrol')
def server_patrol_output():
    if os.path.exists(patrol_image):
        return send_file(patrol_image, mimetype="image/jpg")
    else:
        return send_file(error_image, mimetype="image/jpg"), 404
```

```python
if __name__ == '__main__':
    print("Server running on port 5000...")
    app.run(port=5000)
```

### Below: flask for request & response

```python
@app.route('/process_click', methods=['POST'])
def handle_click():
    # 1. Get JSON data sent from the Qt App
    data = request.json
    print(f"Server: Received click event from {data.get('source')}")

    # 2. Generate a response
    # In a real app, you might query a database here.
    response_text = "Server confirmed: Button was clicked at 100% efficiency!"
    
    return jsonify({
        "status": "ok", 
        "label_update": response_text
    })
```
