```python
import sys
from PyQt5.QtWidgets import (QApplication, QWidget, QPushButton, 
                             QVBoxLayout, QHBoxLayout, QLabel)

class CustomLayoutApp(QWidget):
    def __init__(self):
        super().__init__()
        self.initUI()

    def initUI(self):
        # 1. The Main Layout (The outer container)
        # We want the whole window to be a horizontal stack
        main_layout = QHBoxLayout()

        # --- TOP SECTION: Information ---
        info_label = QLabel("Robot Control Panel", self)
        # Add to main layout
        main_layout.addWidget(info_label)

        # --- MIDDLE SECTION: Directional Buttons (Horizontal) ---
        # We create a NEW layout just for this row
        col_layout = QVBoxLayout()
        
        btn_left = QPushButton("Left")
        btn_center = QPushButton("Center")
        btn_right = QPushButton("Right")

        # Add buttons to the horizontal row
        row_layout.addWidget(btn_left)
        row_layout.addWidget(btn_center)
        row_layout.addWidget(btn_right)

        # IMPORTANT: Add the row layout to the main layout
        main_layout.addLayout(row_layout)

        # --- BOTTOM SECTION: Action Buttons (Vertical) ---
        # Let's add another vertical stack below the row
        action_layout = QVBoxLayout()
        
        btn_fire = QPushButton("FIRE LASER")
        btn_reset = QPushButton("Emergency Reset")
        
        action_layout.addWidget(btn_fire)
        action_layout.addWidget(btn_reset)
        
        # Add this stack to the main layout
        main_layout.addLayout(action_layout)

        # --- OPTIONAL: Spacers ---
        # 'addStretch' pushes everything up by adding empty space at the bottom
        main_layout.addStretch()

        # 2. Apply the Main Layout to the Window
        self.setLayout(main_layout)
        
        self.setWindowTitle('Complex Layouts')
        self.resize(300, 200)

if __name__ == '__main__':
    app = QApplication(sys.argv)
    ex = CustomLayoutApp()
    ex.show()
    sys.exit(app.exec_())
```
